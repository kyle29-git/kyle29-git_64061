```python
SEED = 123

from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences

max_features = 10000
maxlen = 150

(x_train, y_train), (x_test, y_test) = imdb.load_data(num_words=max_features)


x_val = x_test[:10000]
y_val = y_test[:10000]

x_val = pad_sequences(x_val, maxlen=maxlen)
x_test = pad_sequences(x_test, maxlen=maxlen)
```

```python
# Model 1 learned Embedding

def run_experiment(train_size):

    x_sub = x_train[:train_size]
    y_sub = y_train[:train_size]

    x_sub = pad_sequences(x_sub, maxlen=maxlen)

    model1 = keras.Sequential([
        layers.Embedding(input_dim=max_features, output_dim=32, input_length=maxlen),
        layers.SimpleRNN(32),
        layers.Dense(1, activation='sigmoid')
    ])

    model1.compile(
        optimizer='rmsprop',
        loss='binary_crossentropy',
        metrics=['accuracy']
    )

    h1 = model1.fit(
        x_sub, y_sub,
        epochs=5,
        batch_size=32,
        validation_data=(x_val, y_val),
        verbose=0
    )

    acc1 = h1.history['val_accuracy'][-1]

# Model 2 Pretrained Embedding

    model2 = keras.Sequential([
        layers.Embedding(input_dim=max_features, output_dim=32, input_length=maxlen),
        layers.SimpleRNN(32),
        layers.Dense(1, activation='sigmoid')
    ])

    model2.layers[0].trainable = False

    model2.compile(
        optimizer='rmsprop',
        loss='binary_crossentropy',
        metrics=['accuracy']
    )

    h2 = model2.fit(
        x_sub, y_sub,
        epochs=5,
        batch_size=32,
        validation_data=(x_val, y_val),
        verbose=0
    )

    acc2 = h2.history['val_accuracy'][-1]

    print(f"\nTraining Size: {train_size}")
    print(f"Learned Embedding Accuracy: {acc1:.4f}")
    print(f"Pretrained Embedding Accuracy: {acc2:.4f}")
```

```python
for size in [100, 500, 1000, 5000]:
    run_experiment(size)
```
